## Data validation

```python
import pandas as pd
df = pd.read_csv(r"C:\Users\Pdhar\OneDrive\Documents\GitHub\Online_Retail_Transaction_Analysis\Dataset\RawData\Online Retail.csv")
print("Row:", df.shape[0])
print("Columns:", df.shape[1])
```

### InvoiceDate Validation

The 'InvoiceDate'column is check to makesure that transaction dates are store in a valid date format.
Correct date is more important for time analysis.

---

```python
df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"], 
    errors="coerce"
)
print("InvoiceDate type:",
      df["InvoiceDate"].dtype)
print("Invalid dates:",
      df["InvoiceDate"].isna().sum())
```

### Result for InvoiceDate Validation

The InvoiceDate column was successfully change into datetime. It found 0 invalid dates are in a valid format and ready for time base analysis.

## CustomerID Validation

This column is check for missing values because a valid customer identifier is important for customer level analysis and segmentation.

```python
print("Missing CustomerID:", 
df["CustomerID"].isna().sum())
```

### Customer ID Validation Result
    The customer id column was checked for missing values but no missing values were found in the dataset.

## Transaction Invoice Validation

   The 'InvoiceNo' column is checked for missing and cancelled transaction records. This helps ensure that each transaction has a valid invoice number before data cleaning.

```python
print("Missing InvoiceNo:",)
df["InvoiceNo"].isna().sum()
print("Cancelled transactions:",
df["InvoiceNo"].astype(str).str.startswith("c").sum())
```

### Result of Transaction Invoice Validation
    This column was checked for missing and cancelled transactions. Missing invoice were not found. Cancelled transactions were identified and will be considered during the data cleaning process.

### Invalid Transactions Remove
    Invalid transactions removed from dataset including cancelled transactions, negative quantities and zero/negative unit prices.

```python
df 
df[df["InvoiceNo"].astype(str).str.startswith("C")]

print("Rows after removing cancelled transactions:", len(df))
```

```python
df = df[df["Quantity"] > 0]

print("Rows after removing invalid quantities:", len(df))
```

```python
df = df[df["UnitPrice"] > 0]

print("Rows after removing invalid unit prices:", len(df))
```

### Data cleaning result

* Cancelled transactions were removed
* Invalid quantities were removed
* Invalid unitprice were removed.
* Final dataset contains 526,054 rows.

---

## Data Cleaning
   The identified data quality issues are cleaned to prepare the dataset for analysis. Duplicate records, invalid quantities, invalid unit prices and cancelled transactions removed.

```python
df = df.drop_duplicates()
print("Rows after removing duplicates:", len(df))
```

### Duplicate Removal Result
    Duplicate records have been removed from the dataset. The dataset is ready for further cleaning.

## Create new feature or feature engineering

New feature create from the cleaned dataset to support sales analysis.

The TotalAmount feature is calculated by multiplying quantity by Unitprice

```python
# Create TotalAmount feature
df["TotalAmount"] = df["Quantity"] * df["UnitPrice"]

print("TotalAmount feature created successfully.")
print(df[["Quantity", "UnitPrice","TotalAmount"]].head())
```

### New Feature result
The TotalAmount feature was created successfully. This feature represents the total value of each transaction and can be used for further sales analysis.
